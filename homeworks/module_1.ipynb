```python
# Импортируйте все необходимые библиотеки и модули которые вам необходимы
import pandas as pd
```

```python
# Загрузите датасет и проведите его очистку и форматирование если это необходимо
df = pd.read_csv('baseball_games.csv')

print(f"Размер начального датат сета: {df.shape}")

text_columns = df.select_dtypes(include=['object']).columns
for col in text_columns:
    df[col] = df[col].str.replace(r'\s+', ' ', regex=True).str.strip()

print(df.duplicated().sum())
if df.duplicated().sum() > 0:
    df.drop_duplicates()

print(df.isna().sum() > 0)

df['attendance'] = df['attendance'].fillna(df['attendance'].median())

print(f"Размер датат сета: {df.shape}")


```

# 19. Правда ли что большинство проигрышных домашних матчей приходятся на Субботу и Воскресенье?

```python
home_losses = df[df['home_team_loss'] == 1]

losses_by_days = home_losses['day_of_week'].value_counts()

print(losses_by_days[['Saturday', 'Sunday']].sum() / losses_by_days.sum())

# Утверждение ложно
```

Утверждение ложно.

Логика: 
1) Отфильтрованы все матчи, где зафиксировано поражение домашней команды
2) Подсчитано распределение таких матчей по дням недели
3) Суммарная доля домашних поражений, приходящихся на субботу и воскресенье, составляет около 31.52%, что значительно меньше половины от общего числа проигранных дома матчей. Следовательно, большинство поражений приходится на будние дни.

# 20. Правда ли что наибольшее количество ранов происходит в холодную погоду? (Холодной погодой считается погода ниже 0 градусов)

```python
cold_runs = df[df['temperature'] < 32]
warm_runs = df[df['temperature'] >= 32]

print(f"Среднее число очков в холодную погоду: {cold_runs['total_runs'].mean()}")
print(f"Среднее число очков в теплую погоду: {warm_runs['total_runs'].mean()}")

print(cold_runs['total_runs'].sum() / df['total_runs'].sum())
print(warm_runs['total_runs'].sum() / df['total_runs'].sum())
```

Утверждение ложно.

Логика: 
1) Разделяем матчи на две части по условию температуры
2) Затем для корректного сравнения результативности мы используем расчет среднего значения температур
3) Отмечаем, что хотя физически в холодную погоду мяч летит хуже и результативность падает, подавляющее большинство матчей в датасете приходится на теплую погоду, что и объясняет распределение.
