```python
# Импортируйте все необходимые библиотеки и модули которые вам необходимы
import pandas as pd
```

```python
# Загрузите датасет и проведите его очистку и форматирование если это необходимо
df = pd.read_csv('baseball_games.csv')

print(f"Размер начального датасета: {df.shape}")

text_columns = df.select_dtypes(include=['object', 'str']).columns
for col in text_columns:
    df[col] = df[col].str.replace(r'\s+', ' ', regex=True).str.strip()

print(f"Количество дубликатов: {df.duplicated().sum()}")
if df.duplicated().sum() > 0:
    df = df.drop_duplicates()

print(df.isna().sum()[df.isna().sum() > 0])

df['attendance'] = df['attendance'].fillna(df['attendance'].median())

print(f"Размер датасета: {df.shape}")
```

Загружаем исходный датасет, очищаем текстовые поля от лишних пробелов с помощью строковых методов и проверяем таблицу на наличие полных дубликатов. Обнаруженные пропуски в колонке посещаемости заполняем медианным значением, так как медиана устойчива к выбросам и не искажает общее распределение зрителей.

# 18. Зависит ли выигрыш от количества посетителей матча? (https://pandas.pydata.org/docs/reference/api/pandas.DataFrame.cov.html)

```python
covariance_res = df['home_team_win'].cov(df['attendance'])
print(f"Ковариация: {covariance_res}")

correlation_res = df['home_team_win'].corr(df['attendance'])
print(f"Корреляция: {correlation_res}")
```

Рассчитываем ковариацию и линейный коэффициент корреляции между фактом победы домашней команды и числом зрителей на стадионе. Статистически значимой зависимости нет. Полученная ковариация около 224.7 трудно интерпретируется из-за разных масштабов величин, а коэффициент корреляции составляет около 0.0456, что практически равно нулю и доказывает отсутствие линейной связи.

# 20. Правда ли что наибольшее количество ранов происходит в холодную погоду? (Холодной погодой считается погода ниже 0 градусов)

```python
cold_runs = df[df['temperature'] < 32]
warm_runs = df[df['temperature'] >= 32]

print(f"Среднее число очков в холодную погоду: {cold_runs['total_runs'].mean()}")
print(f"Среднее число очков в теплую погоду: {warm_runs['total_runs'].mean()}")

print(cold_runs['total_runs'].sum() / df['total_runs'].sum())
print(warm_runs['total_runs'].sum() / df['total_runs'].sum())
```

Разделяем матчи на две группы по температурному порогу в 32 градуса Фаренгейта, что соответствует нулю градусов Цельсия, и сравниваем среднюю результативность и общую долю очков в каждой группе. Утверждение ложно, так как хотя в выборке при температуре ниже 32 градусов Фаренгейта среднее количество очков выше и составляет 12.0 против 8.93, доля таких холодных матчей в датасете составляет менее 0.06 процента, тогда как 99.94 процента всех ранов за сезон зарабатывается в теплую погоду из-за естественного климатического расписания турнира.
